```python
# Define the discrete-time index
n = np.arange(-5, 11)

# Generate four elementary signals
delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

# Create four subplots
fig, axes = plt.subplots(1, 4, figsize=(15, 3))

signals = [delta, step, ramp, expo]
titles = ["Impulse", "Step", "Ramp", "Exponential"]

for ax, x, title in zip(axes, signals, titles):
    ax.stem(n, x)
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
n = np.arange(-5, 11)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, a in zip(axes, a_values):
    x = (a**n) * (n >= 0)
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

fig, axes = plt.subplots(2, 1, figsize=(10, 7))

axes[0].stem(n, x1)
axes[0].set_title("Cosine Signal: Period 8")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Amplitude")
axes[0].grid(True)

axes[1].stem(n, x2)
axes[1].set_title("Cosine Signal: Period 15")
axes[1].set_xlabel("n")
axes[1].set_ylabel("Amplitude")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
from scipy.io import wavfile
import numpy as np
import matplotlib.pyplot as plt
import os

filename = "my voice.wav"

# Check whether the audio file exists
if not os.path.exists(filename):
    raise FileNotFoundError("Please upload my voice.wav first.")

# Read the WAV file
fs, frame = wavfile.read(filename)

# Convert stereo audio to mono if necessary
if frame.ndim > 1:
    frame = frame.mean(axis=1)

# Normalize the audio
if np.issubdtype(frame.dtype, np.integer):
    info = np.iinfo(frame.dtype)
    x = frame.astype(float) / max(abs(info.min), info.max)
else:
    x = frame.astype(float)

peak = np.max(np.abs(x))
if peak > 0:
    x = x / peak

# Calculate time and duration
t = np.arange(len(x)) / fs
duration = len(x) / fs

print("Sampling rate:", fs, "Hz")
print("Duration:", round(duration, 2), "seconds")
print("Number of samples:", len(x))

# Plot the audio signal
plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("My Recorded Voice Signal")
plt.grid(True)
plt.show()
```
